```python
import nbformat

# Load the notebook file
with open('/content/Lab9_BayesianNetwork.ipynb', 'r', encoding='utf-8') as f:
    nb = nbformat.read(f, as_version=4)

# Print structure and contents of the cells
for idx, cell in enumerate(nb.cells):
    print(f"--- Cell {idx} ({cell.cell_type}) ---")
    print(cell.source[:500] + ('\n...' if len(cell.source) > 500 else ''))
    print('\n')
```

```python
from nbconvert.preprocessors import ExecutePreprocessor
import nbformat

# Load the notebook
with open('/content/Lab9_BayesianNetwork.ipynb', 'r', encoding='utf-8') as f:
    nb = nbformat.read(f, as_version=4)

# Configure and run the execution preprocessor
ep = ExecutePreprocessor(timeout=600, kernel_name='python3')
try:
    print("Starting notebook execution...")
    ep.preprocess(nb, {'metadata': {'path': '/content/'}})
    print("Execution completed successfully! Saving notebook...")

    # Save the executed notebook back
    with open('/content/Lab9_BayesianNetwork.ipynb', 'w', encoding='utf-8') as f:
        nbformat.write(nb, f)

    # Print output of key execution cells
    for idx, cell in enumerate(nb.cells):
        if cell.cell_type == 'code':
            print(f"\n--- [Cell {idx}] Output ---")
            for output in cell.get('outputs', []):
                if 'text' in output:
                    print(output['text'].strip())
                elif 'data' in output and 'text/plain' in output['data']:
                    print(output['data']['text/plain'].strip())
except Exception as e:
    print(f"Error executing notebook: {e}")
```
