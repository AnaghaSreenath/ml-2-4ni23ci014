## Supervised Classification using Bagging and Boosting

This notebook demonstrates how to apply and compare bagging (Random Forest) and boosting (Gradient Boosting) techniques for a classification task using a synthetic dataset.

```python
# Import necessary libraries
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report
import pandas as pd
import numpy as np

print("Libraries imported successfully.")
```

```python
# Generate a synthetic dataset  suitable for binary classification using make_classification.


X, y = make_classification(
    n_samples=1000,          # Total number of samples
    n_features=20,           # Total number of features
    n_informative=10,        # Number of informative features
    n_redundant=5,           # Number of redundant features
    n_repeated=2,            # Number of repeated features
    n_classes=2,             # Number of target classes
    random_state=42          # For reproducibility
)

# Convert to DataFrame for better visualization and manipulation (optional)
df = pd.DataFrame(X, columns=[f'feature_{i}' for i in range(X.shape[1])])
df['target'] = y

print(f"Dataset shape: {df.shape}")
display(df.head())
```

```python
# Split the dataset into training and testing sets
#We'll split the dataset to train models on one part and evaluate them on unseen data.
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

print(f"Training set shape: {X_train.shape}, {y_train.shape}")
print(f"Testing set shape: {X_test.shape}, {y_test.shape}")
```

### 3. Implement Bagging (Random Forest) Classifier

Bagging involves training multiple models (usually of the same type) on different subsets of the training data and averaging their predictions. Random Forest is a popular bagging algorithm for decision trees.

```python
# Initialize and train the Random Forest Classifier
rf_classifier = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
print("Training Random Forest Classifier...")
rf_classifier.fit(X_train, y_train)
print("Random Forest Classifier trained.")

# Make predictions on the test set
y_pred_rf = rf_classifier.predict(X_test)

# Evaluate the Random Forest model
print("\n--- Random Forest Classifier Performance ---")
print(f"Accuracy: {accuracy_score(y_test, y_pred_rf):.4f}")
print(f"Precision: {precision_score(y_test, y_pred_rf):.4f}")
print(f"Recall: {recall_score(y_test, y_pred_rf):.4f}")
print(f"F1-Score: {f1_score(y_test, y_pred_rf):.4f}")
print("\nClassification Report:\n", classification_report(y_test, y_pred_rf))
```

### 4. Implement Boosting (Gradient Boosting) Classifier

Boosting sequentially builds models, where each new model attempts to correct the errors of the previous ones. Gradient Boosting is a powerful boosting algorithm.

```python
# Initialize and train the Gradient Boosting Classifier
gb_classifier = GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, random_state=42)
print("Training Gradient Boosting Classifier...")
gb_classifier.fit(X_train, y_train)
print("Gradient Boosting Classifier trained.")

# Make predictions on the test set
y_pred_gb = gb_classifier.predict(X_test)

# Evaluate the Gradient Boosting model
print("\n--- Gradient Boosting Classifier Performance ---")
print(f"Accuracy: {accuracy_score(y_test, y_pred_gb):.4f}")
print(f"Precision: {precision_score(y_test, y_pred_gb):.4f}")
print(f"Recall: {recall_score(y_test, y_pred_gb):.4f}")
print(f"F1-Score: {f1_score(y_test, y_pred_gb):.4f}")
print("\nClassification Report:\n", classification_report(y_test, y_pred_gb))
```

### 5. Compare Model Performance

Let's summarize and compare the performance metrics of both models.

```python
# Store results in a dictionary
results = {
    'Model': ['Random Forest (Bagging)', 'Gradient Boosting (Boosting)'],
    'Accuracy': [accuracy_score(y_test, y_pred_rf), accuracy_score(y_test, y_pred_gb)],
    'Precision': [precision_score(y_test, y_pred_rf), precision_score(y_test, y_pred_gb)],
    'Recall': [recall_score(y_test, y_pred_rf), recall_score(y_test, y_pred_gb)],
    'F1-Score': [f1_score(y_test, y_pred_rf), f1_score(y_test, y_pred_gb)]
}

# Create a DataFrame for easy comparison
performance_df = pd.DataFrame(results)

print("\n--- Model Performance Comparison ---")
display(performance_df.round(4))

# Highlight the better performing model for each metric
print("\nObservations:")
for metric in ['Accuracy', 'Precision', 'Recall', 'F1-Score']:
    if performance_df.loc[0, metric] > performance_df.loc[1, metric]:
        print(f"- Random Forest has higher {metric}.")
    elif performance_df.loc[1, metric] > performance_df.loc[0, metric]:
        print(f"- Gradient Boosting has higher {metric}.")
    else:
        print(f"- Both models have similar {metric}.")
```
