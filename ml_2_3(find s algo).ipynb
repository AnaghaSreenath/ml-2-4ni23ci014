```python
def find_s_algorithm(data):
    features = data.iloc[:, :-1].values
    # Convert target values to a consistent case (e.g., capitalize first letter) for comparison
    target = data.iloc[:, -1].astype(str).str.capitalize().values

    h = []
    for i, concept in enumerate(target):
        if concept == 'Yes':
            h = list(features[i])
            break

    if not h:
        return None

    for i, concept in enumerate(target):
        if concept == 'Yes':
            for j in range(len(h)):
                if h[j] == '?':
                    continue
                elif h[j] != features[i][j]:
                    h[j] = '?'

    return h
```

```python
data = pd.DataFrame([
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
], columns=['Sky', 'AirTemp', 'Humidity', 'Wind', 'Water', 'Forecast', 'EnjoySport'])

display(data)

specific_hypothesis = find_s_algorithm(data)

if specific_hypothesis:
    print(f"Most Specific Hypothesis (h): {specific_hypothesis}")
```

```python
new_data = pd.read_csv('/content/Sky,AirTemp,Humidity,Wind,Water,For.txt')

print("New Dataset:")
display(new_data)

new_specific_hypothesis = find_s_algorithm(new_data)

if new_specific_hypothesis:
    print(f"Most Specific Hypothesis (h) for new dataset: {new_specific_hypothesis}")
```

```python
import pandas as pd
import numpy as np
df = pd.read_csv("EnjoySport.csv.csv")
X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values
h = np.array(["Ø"] * X.shape[1])
for x, target in zip(X, y):
    if target == "Yes":
        h = np.where(h == "Ø", x,
             np.where(h == x, h, "?"))
print("Final Hypothesis:", h)
```
